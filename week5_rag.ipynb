```python
#build retrieval
from google.cloud import bigquery
bq = bigquery.Client(project="resume-radar-thuan")
```

```python
query = r"""
with target as (
    select embedding
    from resume_radar.posting_embeddings
    where job_id = 3894546413
)
select concat(p.title , " " , p.description) as posting_text, p.job_id, ml.distance(t.embedding, e.embedding, 'cosine') as cos_dist
from resume_radar.posting_embeddings e
cross join target t
join resume_radar.postings p
on p.job_id = e.job_id
where p.job_id != 3894546413
qualify row_number() over(
    partition by COALESCE(p.company_name, CAST(p.job_id AS STRING)), p.title
    order by cos_dist
) = 1
order by cos_dist ASC
limit 5
"""
```

```python
job_postings = []
for row in bq.query(query).result():
    job_postings.append(dict(row))
```

```python
job_postings
```

```python
recommendation = {}
error = {}
for i in range(len(job_postings)):
    try:
        prompt = f"""act as a senior hiring manager and look at this job postings {job_postings[i]['posting_text']} can compare it with the resume {resume_text}."""
        resp = client.models.generate_content(
        model = "gemini-2.5-flash-lite",
        contents = prompt,
        config={
            "response_mime_type" : "application/json",
            "response_schema" : FitAnalysis,
        },
        )
        fit = FitAnalysis.model_validate_json(resp.text)
        
        recommendation[job_postings[i]['job_id']] = fit

    except Exception as e:
        error[job_postings[i]['job_id']] = str(e)
```

```python
print(recommendation)
```

```python
def check_grounding(fit, posting_text, resume_text):
    violations = []
    posting_text_lower = posting_text.lower()
    resume_text_lower = resume_text.lower()
    for skill in fit.matched_skills:
        if not contains(skill, posting_text) or not contains(skill, resume_text):
            violations.append({"matched" : skill})
    for skill in fit.gap_skills:
        if not contains(skill, posting_text):
            violations.append({"gap not in posting" :  skill})
        if contains(skill, resume_text):
            violations.append({"gap in resume" :  skill})
    return violations
```

```python
import re

def contains(skill, text):
    pattern = r'\b' + re.escape(skill) + r'\b'
    return re.search(pattern, text, re.IGNORECASE) is not None
```

```python
posting_lookup = {}
for row in job_postings:
    posting_lookup[row['job_id']] = row['posting_text']  

for job_id, fit in recommendation.items():
    posting_text = posting_lookup[job_id]
    v = check_grounding(fit, posting_text, resume_text)
    print(job_id, v)
```
